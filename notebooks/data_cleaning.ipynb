```python
import pandas as pd
```

```python
def clean_data(year):
    df = pd.read_excel(f"../data/raw/{year}년_식중독.xls")
    df = df.drop(columns=['합계', 'Unnamed: 9'])
    df = df[df['월'] != '합계']
    df.insert(0,'연도',year)
    df['월'] = df['월'].astype(int)
    df = df.melt(
    id_vars=['연도', '월', '구분'],
    var_name='시설',
    value_name='값')

    df = df.pivot(
    index=['연도', '월', '시설'],
    columns='구분',
    values='값'
    ).reset_index()
    return df
```
